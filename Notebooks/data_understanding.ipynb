## **Project Strategy: Core Metrics & Business Questions**
**Goal:** Analyze the banking dataset to identify the primary drivers of customer attrition (Churn) and provide data-driven retention strategies.

## 1. Core Project Metrics
We will focus on these foundational metrics to ensure a clear and accurate business story:
| **Metric** | **Purpose / Column Used** |
| :-----: | :-----: |
| **Total Churn Rate** | Measures the overall percentage of customers lost. (Exited) |
| **Geographic Risk** | Identifies which regions are `churn hotspots`. (Geography, Exited) |
| **Financial Vulnerability** | Evaluates if account balance size influences loyalty. (Balance, Exited) |
| **Engagement Impact** | Shows the correlation between bank activity and retention. (IsActiveMember, Exited) |
| **Credit Worthiness** | Analyzes if credit scores impact the decision to leave. (CreditScore, Exited) |
## 2. Key Analysis Questions
We will answer these specific questions to derive actionable business recommendations:
1. **Regional Impact:** Is the churn rate significantly higher in one specific country (e.g., Germany vs. France)?
2. **Financial Threshold:** Are low-balance customers more likely to churn, or is the bank losing high-value VIP clients?
3. **Product Stickiness:** Does owning multiple products (Credit Cards, etc.) increase a customer's "stickiness" to the bank?
4. **Member Activity:** Can increasing the engagement of "Inactive Members" effectively reduce churn?
5. **Demographic Trends:** Which age groups or genders show the highest risk of leaving the bank?
