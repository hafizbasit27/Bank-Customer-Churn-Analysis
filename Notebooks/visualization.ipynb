# **Phase 5: Visualizations**

```python
# Loading data in Python for processing.
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('cleaned_dataset_for_visualization.csv')
df.head()
```

```python
# Counting the number of customers who stayed and exited
churn_counts = df['Exited'].value_counts()
# Creating the Pie Chart
plt.figure(figsize=(8, 6))
plt.pie(churn_counts, labels=['Stayed (0)', 'Exited (1)'], autopct='%1.1f%%', colors=['green', 'red'], startangle=140)
plt.title('Overall Customer Churn Distribution')
plt.show()
```

```python
# Creating the cross-tabulation
geo_churn = pd.crosstab(df['Geography'], df['Exited'])

# Plotting with 'ax' to handle labels
ax = geo_churn.plot(kind='bar', figsize=(10, 6), color=['green', 'red'])

# Adding Data Labels
for container in ax.containers:
     ax.bar_label(container, padding=3)
# Decoration
plt.title('Churn Count by Geography with Labels')
plt.xlabel('Country')
plt.ylabel('Number of Customers')
plt.legend(['Stayed', 'Exited'])
plt.xticks(rotation=0)
plt.show()
```

```python
# Filtering data for exited customers only
exited_only = df[df['Exited'] == 1]
plt.figure(figsize=(10, 6))
# Plotting the histogram
plt.hist(exited_only['Age'], bins=25, color='red', edgecolor='white')
# Adding labels and title
plt.title('Distribution of Customers Who Left (Age-Wise)')
plt.xlabel('Age of Exited Customers')
plt.ylabel('Number of Customers')
# Optional: Adding a grid for better readability
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
# Filtering data for exited customers only
exited_only = df[df['Exited'] == 1]
plt.figure(figsize=(10, 6))
# Plotting the histogram for Balance
plt.hist(exited_only['Balance'], bins=30, color='orange', edgecolor='black')
# Adding labels and title
plt.title('Balance Distribution of Customers Who Left')
plt.xlabel('Account Balance')
plt.ylabel('Number of Customers')
plt.show()
```

```python
# Grouping by Number of Products and Exited
prod_churn = pd.crosstab(df['NumOfProducts'], df['Exited'])
# Plotting
ax = prod_churn.plot(kind='bar', figsize=(10, 6), color=['green', 'red'])
# Adding Labels
for container in ax.containers:
    ax.bar_label(container, padding=3)
plt.title('Churn Count by Number of Products')
plt.xlabel('Number of Products')
plt.ylabel('Customer Count')
plt.legend(['Stayed', 'Exited'])
plt.xticks(rotation=0)
plt.show()
```

```python
# Grouping by Activity and Exited
active_churn = pd.crosstab(df['IsActiveMember'], df['Exited'])
# Plotting
ax = active_churn.plot(kind='bar', figsize=(8, 6), color=['green', 'red'])
# Adding Labels
for container in ax.containers:
    ax.bar_label(container, padding=3)
plt.title('Impact of Being an Active Member on Churn')
plt.xlabel('Is Active Member (0 = No, 1 = Yes)')
plt.ylabel('Customer Count')
plt.legend(['Stayed', 'Exited'])
plt.xticks(rotation=0)
plt.show()
```
