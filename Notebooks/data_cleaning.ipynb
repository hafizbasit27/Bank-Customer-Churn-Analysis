## **Data Cleaning & Refinement**
In this phase, we refined the raw dataset to ensure it is optimized for exploratory analysis.
## 1. Feature Selection (Removing Noise)
* **Action:** We dropped the following columns: `RowNumber`, `CustomerId`, and `Surname`.
* **Business Rationale:** These columns are personal identifiers or system-generated indices. Since they do not possess any predictive power regarding a customer's decision to leave the bank, removing them reduces the dimensionality of our data and prevents `overfitting` during analysis.
## 2. Handling Indexing Artifacts
* **Action:** Removed the Unnamed: 0 column which appeared to be a remnant of a previous CSV save operation.
## 3. Validation of Data Types
* **Status:** Performed a check on column formats. All critical metrics such as `CreditScore`, `Balance`, and `Age` are correctly identified as `numerical (int/float)`, ensuring that mathematical operations in the EDA phase will be accurate.
