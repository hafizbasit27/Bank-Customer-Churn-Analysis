# **Phase 4: Exploratory Data Analysis (EDA)**

```python
# Loading file in Python for processing.
import pandas as pd
df = pd.read_csv('cleaned_dataset_for_EDA.csv')
df.head()
```

```python
df = df.drop(['Unnamed: 0'], axis= 1)
df.head()
```

```python
# Calculating overall churn percentage
overall_churn = df['Exited'].value_counts(normalize=True) * 100
print("Overall Churn Rate (0 = Stayed, 1 = Exited):")
print(overall_churn)
```

```python
# Regional churn distribution
region_churn = pd.crosstab(df['Geography'], df['Exited'], normalize='index') * 100
print("Churn Percentage by Geography:")
print(region_churn)
```

```python
# Comparing average Balance and Salary for churned vs stayed customers
financial_analysis = df.groupby('Exited')[['Balance', 'EstimatedSalary']].mean()
print("Average Financial Metrics by Exited Status:")
print(financial_analysis)
```

```python
# Product distribution vs Churn
product_churn = pd.crosstab(df['NumOfProducts'], df['Exited'], normalize='index') * 100
print("Churn Rate by Number of Products:")
print(product_churn)
```

```python
# Activity status vs Churn
active_churn = pd.crosstab(df['IsActiveMember'], df['Exited'], normalize='index') * 100
print("Churn Rate: Inactive (0) vs Active (1) Members:")
print(active_churn)
```

```python
# Average Age of customers who stayed vs those who left
age_analysis = df.groupby('Exited')['Age'].mean()
print("Average Age by Exit Status:")
print(age_analysis)
# Gender-based churn rate
gender_churn = pd.crosstab(df['Gender'], df['Exited'], normalize='index') * 100
print("Churn Rate by Gender:")
print(gender_churn)
```

## **Exploratory Data Analysis (EDA)**
In this phase, we conducted a numerical investigation to identify patterns and relationships within the customer data without visual aids.
## 1. Global Attrition Overview
* **Objective:** Determine the baseline churn rate for the entire customer base.
* **Metric:** We calculated the percentage distribution of the Exited column to understand the scale of customer loss.
## 2. Geographical Risk Assessment
* **Objective:** Identify `High-Risk` regions.
* **Logic:** By cross-tabulating Geography with Exited status, we identified which countries are experiencing the highest percentage of customer departures. This helps the bank prioritize regional retention strategies.
## 3. Financial Correlation Analysis
* **Objective:** Analyze if customer wealth (Balance/Salary) influences loyalty.
* **Logic:** We aggregated the average Balance and EstimatedSalary for both groups (Exited vs. Stayed). This reveals if the bank is losing high-value (VIP) customers or those with lower engagement levels.
## 4. Product Portfolio Stickiness
* **Objective:** Analyze if `Product Bundling` (owning multiple bank products) increases customer loyalty.
* **Metric:** Cross-tabulation of `NumOfProducts` against Exited status to find the optimal number of products for retention.
## 5. Behavioral Engagement Analysis
* **Objective:** Validate the impact of the `IsActiveMember` flag on retention.
* **Observation:** By comparing active vs. inactive members, we quantify how much the bank's engagement programs contribute to lowering churn rates.
## 6. Demographic Risk Profiling
* **Objective:** Identify `high-risk` segments based on `Age` and `Gender`.
* **Logic:** Calculated average age and gender-wise percentages to see if the bank is losing a specific generation or demographic group.

```python
# Select only numerical columns for correlation
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns
correlation_matrix = df[numerical_cols].corr()

print("Numerical Correlation with Exited Status:")
print(correlation_matrix['Exited'].sort_values(ascending=False))
```

## Statistical Correlation Findings
Based on the numerical correlation matrix, we have identified the primary statistical drivers for customer attrition:

* **Primary Driver - Age:** Shows a positive correlation of `0.28`, suggesting that older customers are more likely to exit the bank.
* **Financial Impact - Balance:** A correlation of `0.11` indicates that churn is slightly higher among customers with larger account balances, signifying a loss of high-value clients.
* **Retention Factor - Activity:** `IsActiveMember` has a negative correlation `-0.15`, confirming that active engagement is a key defense against churn.
* **Product Stickiness:** `NumOfProducts` also shows a negative trend, supporting the theory that multi-product customers are more loyal.
